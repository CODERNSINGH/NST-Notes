### Verifying Normal Distribution for `sensor_noise`

Since `np.random.randn()` generates samples from a standard normal distribution (mean 0, variance 1), we can verify this empirically using:

1.  **Histograms**: To visually inspect the shape of the distribution.
2.  **Q-Q Plots**: To compare the quantiles of the data against the quantiles of a theoretical normal distribution.
3.  **Statistical Tests**: Such as the Shapiro-Wilk test for normality.

### Question

Create the following matrices using NumPy.

**Matrix A (3×4)**

```

**Matrix B (4×3)**

```
1  2  3

Perform the following:
1. Create Matrix A and Matrix B.
2. Display both matrices.
3. Display shape, dimensions and size.
4. Perform matrix multiplication using `np.dot()` and `np.matmul()`.
5. Display transpose using both `.T` and `np.transpose()`.
6. Perform element-wise addition, subtraction, multiplication and division between **A** and **B.T**.


````python
#Example code to check how numpy array is faster than traditional python list

import numpy as np
import time

n = 100000000
time_start = time.time()
l = [ i + 1 for i in range(n)]
time_end = time.time()
print (f"Total time list took : {time_end-time_start}seconds")

time_start = time.time()
l = np.arange(1,n+1)
time_end = time.time()
print (f"Total time numpy took : {time_end-time_start}seconds")
````

````python
# Create Matrix A using arange() and reshape()

A = np.arange(1, 13).reshape(3, 4)
A[:2, 1:2] = 1
A[:, 1] = 0

# Create Matrix B
B = np.arange(1, 13).reshape(4, 3)

print("Matrix A")
print(A)

print("\nMatrix B")
print(B)

# Matrix Information
print("\n--- Matrix A Information ---")
print("Shape:", A.shape)
print("Dimensions:", A.ndim)
print("Size:", A.size)

print("\n--- Matrix B Information ---")
print("Shape:", B.shape)
print("Dimensions:", B.ndim)
print("Size:", B.size)

# Matrix Multiplication
print("\nnp.dot(A, B)")
print(np.dot(A, B))

print("\nnp.matmul(A, B)")
print(np.matmul(A, B))

# Transpose
print("\nA.T")
print(A.T)

print("\nnp.transpose(B)")
print(np.transpose(B))

# Element-wise Operations (A and B.T have same shape)
print("\nElement-wise Addition")
print(A + B.T)

print("\nElement-wise Subtraction")
print(A - B.T)

print("\nElement-wise Multiplication")
print(A * B.T)

print("\nElement-wise Division")
print(A / B.T)
````

## Part A: Create a Sample Student Performance Dataset

> If your instructor provides a CSV/Excel file, use `pd.read_csv()` or `pd.read_excel()`.


````python
import pandas as pd
# Create sample dataset
students = {
    "StudentID":[101,102,103,104,105,106,107,108],
    "Name":["Aman","Riya","Karan","Sneha","Rahul","Priya","Arjun","Neha"],
    "Department":["CSE","CSE","AI","AI","CSE","AI","CSE","AI"],
    "Math":[85,91,76,88,65,95,72,81],
    "Science":[80,89,70,90,68,92,75,84],
    "Attendance":[92,96,85,98,80,97,88,91]
}

df = pd.DataFrame(students)

# Save files so students can practice loading
df.to_csv("student_performance.csv", index=False)
df.to_excel("student_performance.xlsx", index=False)

print(df)
````

## Part B: Loading Data


````python
# Load CSV
df_csv = pd.read_csv("student_performance.csv")
print("Loaded from CSV")
display(df_csv)

# Load Excel
df_excel = pd.read_excel("student_performance.xlsx")
print("Loaded from Excel")
display(df_excel)
````

## Part C: Selection Operations

````python
# Single column (Series)
print(df['Math'])
print("\n")
# Multiple columns (DataFrame)
display(df[['Name','Math','Science']])

print("\n")
# loc
print(df.loc[2,'Name'])

print("\n")
# iloc
#  integer-location based indexing
print(df.iloc[2,3])

print("\n")
# iloc
print(df.iloc[2,2:3])

print("\n")
# at
print(df.at[0,'Attendance'])

print("\n")
# iat
print(df.iat[0,5])
````


## Part D: Filtering, Sorting and Grouping

````python
# Boolean Indexing
display(df[df['Math'] > 80])
````

````python
# query()
display(df.query("Attendance >= 90"))
````

````python
# sort_values()
display(df.sort_values(by='Science', ascending=False))
````

````python
# groupby().agg()
summary = df.groupby("Department").agg({
    "Math":"mean",
    "Science":"mean",
    "Attendance":"mean"
})

display(summary)
````


## 4. Dataset Inspection & Quality Checks

````python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nInfo:")
df.info()

print("\nSummary Statistics:")
display(df.describe())
````

````python
print("Unique Departments")
print(df["Department"].unique())

print("\nDepartment Counts")
print(df["Department"].value_counts())

print("\nDepartment Counts")
print(df["Department"].nunique())
````

````python
duplicate_df = pd.concat([df, df.iloc[[2]]], ignore_index=True)

print("Dataset with Duplicate Row")
display(duplicate_df)

print("Duplicate Rows")
display(duplicate_df[duplicate_df.duplicated()])

print("Number of Duplicates:", duplicate_df.duplicated().sum())

duplicate_df.drop_duplicates(inplace=True)

print("\nAfter Removing Duplicates")
display(duplicate_df)
````

````python
print("Logical Validation (Attendance between 0 and 100)")
valid = duplicate_df[(duplicate_df["Attendance"]>=0) &
                     (duplicate_df["Attendance"]<=100)]
display(valid)
````
